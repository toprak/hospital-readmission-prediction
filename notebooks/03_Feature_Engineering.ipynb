```python
import pandas as pd
import numpy as np

# Load the Cleaned Dataset
print("Loading clean data...")
df = pd.read_csv('../data/processed/cleaned_dataset.csv')

# Dropping Columns That Provide No Contribution to the Model
useless_meds = ['acetohexamide', 'tolbutamide', 'miglitol', 'troglitazone', 
                'tolazamide', 'examide', 'citoglipton', 'glimepiride-pioglitazone', 
                'metformin-rosiglitazone', 'metformin-pioglitazone', 'nateglinide', 
                'chlorpropamide', 'acarbose']

df.drop(columns=[col for col in useless_meds if col in df.columns], inplace=True)

# Dropping Useless 'Err' Columns from Demographic Data
err_cols = [col for col in df.columns if 'Err' in col]
df.drop(columns=err_cols, inplace=True)

# Preventing Column Explosion by Dropping High Cardinality Columns
high_cardinality_cols = ['diag_1', 'diag_2', 'diag_3', 'State', 'County', 'CensusTract']
df.drop(columns=[col for col in high_cardinality_cols if col in df.columns], inplace=True)

# Converting the Age Column to Numeric Values
age_mapping = {
    '[0-10)': 5, '[10-20)': 15, '[20-30)': 25, '[30-40)': 35, '[40-50)': 45,
    '[50-60)': 55, '[60-70)': 65, '[70-80)': 75, '[80-90)': 85, '[90-100)': 95
}
df['age'] = df['age'].map(age_mapping)

# Encoding Remaining Categorical Data into Machine Language
print("Categorical data is being encoded...")
categorical_cols = df.select_dtypes(include=['object']).columns
df_encoded = pd.get_dummies(df, columns=categorical_cols, drop_first=True)

# Saving the Engineered Data
final_output_path = '../data/processed/engineered_dataset.csv'
df_encoded.to_csv(final_output_path, index=False)

print("-" * 50)
print("Feature Engineering completed! The data is now fully in machine language (numerical).")
print(f"Original Number of Columns (After Cleaning): {df.shape[1]}")
print(f"New Number of Columns After Encoding: {df_encoded.shape[1]}")
```
