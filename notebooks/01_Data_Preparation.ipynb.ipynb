```python
import pandas as pd
import numpy as np

# Reading the Data
print("Loading data...")
clinical_df = pd.read_csv('../data/raw/diabetic_data.csv')
demographic_df = pd.read_csv('../data/raw/acs2015_census_tract_data.csv')

# Using All the Data
sampled_clinical = clinical_df.copy()

# Foreign Key to Merge the Two Datasets
unique_locations = demographic_df['CensusTract'].unique()
np.random.seed(42)
sampled_clinical['CensusTract'] = np.random.choice(unique_locations, size=len(sampled_clinical))

# Merging the Tables
print("Merging tables...")
custom_dataset = pd.merge(sampled_clinical, demographic_df, on='CensusTract', how='left')

# Saving our Custom Dataset
output_path = '../data/processed/custom_readmission_dataset.csv'
custom_dataset.to_csv(output_path, index=False)

print("-" * 50)
print("All data successfully merged!")
print(f"Total Number of Rows: {custom_dataset.shape[0]}")
print(f"Total Number of Columns (Features): {custom_dataset.shape[1]}")
```
